# NumPy: desafio laranja × toranja

Relação entre diâmetro e peso de laranjas e toranjas, coeficientes de uma regressão linear calculados à mão com NumPy e busca aleatória do coeficiente angular que minimiza o erro.

```python
import numpy as np
```

```python
dado = np.loadtxt('https://raw.githubusercontent.com/allanspadini/numpy/dados/citrus.csv', delimiter=',', usecols=np.arange(1,6,1), skiprows=1)
```

```python
diametro_laranja = dado[:5000,0]
diametro_toranja = dado[5000:,0]
peso_laranja = dado[:5000,1]
peso_toranja = dado[5000:,1]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(diametro_laranja, peso_laranja)
plt.plot(diametro_toranja, peso_toranja)
plt.legend(['laranja', 'toranja'])
plt.xlabel('Diâmetro')  # Nomeia o eixo X
plt.ylabel('Peso')      # Nomeia o eixo Y
```

```python
Y = peso_laranja
X = diametro_laranja
n = np.size(X)
a = (n * np.sum(X * Y) - np.sum(X) * np.sum(Y)) / (n * np.sum(X ** 2) - np.sum(X)** 2)
b = np.mean(Y) - a * np.mean(X)
```

```python
Y = peso_toranja
X = diametro_toranja
n = np.size(X)
a = (n * np.sum(X * Y) - np.sum(X) * np.sum(Y)) / (n * np.sum(X ** 2) - np.sum(X)** 2)
b = np.mean(Y) - a * np.mean(X)
```

```python
norma = np.array([])
b =17
np.random.seed(84)
coef_angulares = np.random.uniform(low=0.0,high=30.0,size=100)

for i in range(100):
  norma = np.append(norma,np.linalg.norm(Y- (coef_angulares[i]*X+b)))
```
